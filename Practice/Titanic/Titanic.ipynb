```python
import seaborn as sns

# Load the built-in Titanic dataset
df = sns.load_dataset('titanic')

# Display the first few rows
print(df.head())
```

```python
# We are predicting whether or not a person survived, so the 'survived' column is the target.
# The 'pclass' column is an integer version of the 'class' column. No need to keep 'class'.
# 
# What is 'sibsp' column?
# Is 'alive' column the same as 'survived' column?
```

```python
import matplotlib.pyplot as plt
import pandas as pd

pd.crosstab(df['sex'], df['survived']).plot(kind='bar', figsize=(8, 5))

plt.title('Titanic Survival by Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No (0)', 'Yes (1)'])
plt.xticks(rotation=0)
plt.show()
```

```python
pd.crosstab(df['age'], df['survived']).plot(kind='bar', figsize=(8, 5))

plt.title('Titanic Survival by Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No (0)', 'Yes (1)'])
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.violinplot(
    data=df,
    x="sex",
    y="age",
    hue="survived",
    split=True,
    inner="quartile",
    palette="muted",
)

# Format the chart
plt.title("Titanic Survival Distribution by Sex and Age", fontsize=14)
plt.xlabel("Sex", fontsize=12)
plt.ylabel("Age", fontsize=12)
plt.legend(title="Survived", labels=["No", "Yes"])
plt.show()
```
